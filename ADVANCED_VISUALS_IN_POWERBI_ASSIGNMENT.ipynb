4. What is a Visual-Level Filter, and how does it differ from Page-Level and Report-Level Filters?

Answer:

-->Visual-Level Filter: Affects only one visual (chart, table, etc.).
-->Page-Level Filter: Affects all visuals on a single page.
-->Report-Level Filter: Affects all visuals across all pages of the report.

Example:
If Region = South is applied as:

-->Visual-Level Filter → only one chart shows South data.
-->Page-Level Filter → all visuals on that page show South data.
-->Report-Level Filter → entire report shows South data.

6)What is the role of Drill-Down in Power BI? How does it improve data exploration?

ANS:-
Drill-Down in Power BI is a feature that allows users to move from a higher-level summary of data to more detailed levels within a hierarchy. It helps users analyze data step by step and gain deeper insights.

For example, in a sales report with the hierarchy:

Country → Region → Product Category → Product

a user can start by viewing total sales by Country and then drill down to Region, Product Category, and finally Product-level sales.

How it improves data exploration:
Helps users investigate data in greater detail.
Makes it easier to identify trends and patterns.
Allows analysis at multiple levels without creating separate reports.
Provides an interactive way to explore and understand business performance.

Example:
A manager can view total sales for India, drill down to a specific region, then to a product category, and finally to individual products to identify which products contribute the most to sales.

10)Explain the difference between Drill-Down and Drill-Through in Power BI using a real-world example.
ANS:-
| Drill-Down                                                                       | Drill-Through                                                          |
| -------------------------------------------------------------------------------- | ---------------------------------------------------------------------- |
| Used to navigate through different levels of a hierarchy within the same visual. | Used to navigate from one report page to another detailed report page. |
| Shows more detailed data step by step.                                           | Shows detailed information related to a selected item.                 |
| Works within the same chart or visual.                                           | Opens a separate report page.                                          |


Real-World Example:

Suppose a sales manager is analyzing company sales.

Drill-Down:
The manager starts with sales by Country, then drills down to Region, and finally to Product Category to understand sales performance at different levels.

Country → Region → Product Category

Drill-Through:
The manager notices that Laptop sales are high and wants more details. By right-clicking on Laptop and selecting Drill Through, Power BI opens a new page showing region-wise sales for Laptop products.

Conclusion:

Drill-Down is used to explore data within the same visual by moving through hierarchy levels, while Drill-Through is used to open a separate detailed report page for deeper analysis of a selected item.

### 13. What is the “Keep All Filters” option in Drill-Through? How does it affect user interaction?

**Answer:**

The **Keep All Filters** option in Power BI Drill-Through ensures that all filters applied on the source page are carried over to the Drill-Through page along with the selected drill-through field.

When enabled, users see detailed information that remains filtered according to their previous selections, providing a more consistent and focused analysis experience.

### Example:

Suppose a user:

* Selects **Region = North**
* Selects **Category = Electronics**
* Then performs a **Drill-Through** on **Laptop**

If **Keep All Filters = ON**:

* The Drill-Through page shows only **Laptop sales in the North region and Electronics category**.

If **Keep All Filters = OFF**:

* The Drill-Through page shows **all Laptop sales**, ignoring the Region and Category filters from the previous page.

### How it affects user interaction:

* Maintains filter context across pages.
* Provides more relevant and personalized analysis.
* Reduces the need to reapply filters on the Drill-Through page.
* Improves report navigation and user experience.

### Result:

The **Keep All Filters** option preserves all active filters during Drill-Through navigation, ensuring users view detailed data within the same filtering context as the source report page. ✅
