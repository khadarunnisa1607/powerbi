1) What is Power BI and why is it used in businesses?
ans:-Microsoft Power BI is a business intelligence and data visualization tool developed by Microsoft. It helps users collect, analyze, visualize, and share data in the form of reports and dashboards.

Why businesses use Power BI:
To make better business decisions using data.
To create interactive dashboards and reports.
To track company performance in real time.
To combine data from different sources into one platform.
To identify trends, patterns, and insights easily.
To improve productivity and data management.

Question 2: Name and explain the three main components of Power BI.
1. Power BI Desktop

A desktop application used to create reports, transform data, and build visualizations.

2. Power BI Service

An online cloud-based service used to publish, share, and collaborate on reports and dashboards.

3. Power BI Mobile

A mobile application that allows users to access reports and dashboards from smartphones and tablets.

Question 3: Explain the Power BI workflow.

The Power BI workflow includes the following steps:

Connect Data – Import data from different sources such as Excel, SQL databases, or websites.
Transform Data – Clean and organize data using Power Query.
Model Data – Create relationships between tables and perform calculations.
Create Reports – Design charts, graphs, and dashboards.
Publish Reports – Upload reports to Power BI Service.
Share and Collaborate – Share dashboards with team members for business insights.

Question 6: Define Data View, Report View, and Model View. Explain the purpose of each view.
1. Data View

Data View displays the data in table format.

Purpose:

To inspect data records
To create calculated columns and measures
To verify imported data

2. Report View

Report View is used to create visual reports and dashboards.

Purpose:

To build charts, graphs, and visuals
To design interactive dashboards
To analyze business insights visually
3. Model View

Model View shows relationships between tables.

Purpose:

To manage table relationships
To create a proper data model
To improve report performance and analysis

Question 8: Split Owner Name to create two new columns as First Name and Last Name.
Steps:
Open Power Query Editor.
Select the Owner Name column.
Go to the Transform tab.
Click Split Column → By Delimiter.
Choose Space as the delimiter.
Select Split at Left-most delimiter.
Click OK.
Rename the new columns as First Name and Last Name.
