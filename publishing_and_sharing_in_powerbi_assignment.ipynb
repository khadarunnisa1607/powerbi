1)Explain why publishing and sharing reports is considered more important than just building dashboards
in Power BI Desktop.


ANS:-
Building dashboards in Power BI Desktop is only the first step. 2)Define Power BI Service and explain its role in comparison to Power BI Desktop in the reporting lifecycle.


ANS:-Power BI Service is a cloud-based platform used to publish, share, collaborate, and manage Power BI reports and dashboards online.

| Power BI Desktop                      | Power BI Service                   |
| ------------------------------------- | ---------------------------------- |
| Used to create reports and dashboards | Used to publish and share reports  |
| Installed on a local computer         | Cloud-based application            |
| Data modeling and visualization       | Collaboration and distribution     |
| Used by report developers             | Used by report consumers and teams |

Role in reporting lifecycle: Power BI Desktop is used for report creation, while Power BI Service is used for publishing, sharing, refreshing, and managing reports.

8) Q8. Why are Power BI Apps considered more secure and scalable than direct sharing? Explain this from a governance and enterprise usage perspective.

Power BI Apps are more secure and scalable because they allow organizations to distribute reports to large groups of users through a centralized platform rather than sharing reports individually.

From a governance perspective:

Access is managed centrally, making it easier to control permissions.
Content is distributed in a standardized and consistent manner.
Reduces the risk of accidental sharing or unauthorized access.
Supports organizational policies and compliance requirements.

From an enterprise usage perspective:

A single App can be shared with hundreds or thousands of users.
Updates made to reports are automatically available to all App users.
Easier maintenance compared to managing individual report-sharing permissions.
Provides a better user experience by grouping related reports and dashboards in one place.

Conclusion: Power BI Apps improve security, governance, scalability, and administrative efficiency for large organizations.



Q9. Explain Row-Level Security (RLS) in Power BI. How does RLS ensure that different users see different data using the same report?

Row-Level Security (RLS) is a Power BI feature that restricts data access at the row level based on user roles and permissions.

How RLS works:

Roles are created in Power BI Desktop.
Filters are applied to restrict data visibility for each role.
Users are assigned to these roles in Power BI Service.
When users open the report, they only see the data allowed by their assigned role.

Example:
Suppose a company has sales data for different regions.

User	Region Visible
Manager A	North Region
Manager B	South Region
CEO	All Regions

All users access the same report, but RLS automatically filters the data so each user sees only the information they are authorized to view.

Benefit: RLS enhances security and privacy while eliminating the need to create separate reports for different users.

Q10. Assume an organization shares reports without using Apps, workspace roles, or RLS. Discuss any three risks or issues related to security, governance, or data misuse, and explain how Power BI best practices help prevent them.
1. Unauthorized Data Access

Risk:
Users may gain access to sensitive information that they should not see.

Prevention:

Implement Row-Level Security (RLS).
Assign appropriate workspace roles.
Follow the principle of least privilege.
2. Poor Governance and Content Management

Risk:
Multiple versions of reports may be shared, causing confusion and inconsistent decision-making.

Prevention:

Use Power BI Apps for controlled distribution.
Manage content through centralized workspaces.
Maintain a single source of truth.
3. Accidental Data Misuse or Modification

Risk:
Users may unintentionally modify, delete, or misuse reports and datasets.

Prevention:

Assign role-based permissions (Admin, Member, Contributor, Viewer).
Restrict editing rights to authorized personnel only.
Monitor and audit workspace activities.

Conclusion:
Without Apps, workspace roles, and RLS, organizations face significant risks related to security, governance, and data integrity. Using Power BI best practices ensures secure access, effective governance, and reliable reporting across the organization.
